##**Tạo đặc trưng (Feature Engineering)**

```python
import pandas as pd
import numpy as np

#Tạo cột Age
df["Birthday"] = pd.to_datetime(df["Birthday"])
df["application_date"] = pd.to_datetime(df["application_date"])

df["Age"] = df["application_date"].dt.year - df["Birthday"].dt.year

df[['application_date', 'Birthday', 'Age']]
```

```python
#Tạo AgeGroup
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 18, 25, 35, 45, 55, 65, np.inf],
    labels=["<18", "18-25", "26-35", "36-45", "46-55", "56-65", "65+"]
)


df[['Age','AgeGroup']]
```

```python
#Tính LoanTermMonths (kỳ hạn vay theo tháng)
import pandas as pd

df["FromDate"] = pd.to_datetime(df["FromDate"])
df["ToDate"] = pd.to_datetime(df["ToDate"])

df["LoanTermMonths"] = (
    (df["ToDate"].dt.year - df["FromDate"].dt.year) * 12 +
    (df["ToDate"].dt.month - df["FromDate"].dt.month))

df[['FromDate', 'ToDate', 'LoanTermMonths']]
```

```python
#Số ngày xử lý hồ sơ
df["ProcessingDays"] = (df["FromDate"] - df["application_date"]).dt.days

df[["application_date", "FromDate", "ProcessingDays"]]
```

```python
#LoanToIncomeRatio giúp đánh giá khả năng trả nợ dựa trên lương của khách hàng
df["LoanToIncomeRatio"] = (df["Số tiền đăng ký vay ban đầu"] / df["Salary"]).round(2)

df[["Số tiền đăng ký vay ban đầu", "Salary", "LoanToIncomeRatio"]]
```

```python
#Thêm cột số tiền còn lại cần giải ngân
df["Cần Giải Ngân"] = df["Số tiền đăng ký vay ban đầu"] - df["Tiền giải ngân"]

df[["Số tiền đăng ký vay ban đầu", "Tiền giải ngân", "Cần Giải Ngân"]]
```

```python
#Tỷ lệ đã trả của khoản vay
df["Đã trả/khoản vay đã giải ngân"] = (df["Tiền gốc còn lại"] / df["Tiền giải ngân"]).round(2)

df[["Tiền giải ngân", "Tiền gốc còn lại", "Đã trả/khoản vay đã giải ngân"]]
```

```python
#Điểm tín dụng thấp
df["LowCreditScore"] = (df["TS_CREDIT_SCORE_V2"] < 600).astype(int)
df[["TS_CREDIT_SCORE_V2", "LowCreditScore"]]
```

```python
#CreditScoreGroup
df["CreditScoreGroup"] = pd.cut(
    df["TS_CREDIT_SCORE_V2"],
    bins=[-np.inf, 600, 700, 800, np.inf],
    labels = ["Low (High Risk)", "Medium", "Good", "Excellent"])

df[['TS_CREDIT_SCORE_V2', 'CreditScoreGroup']]
```

```python
#IncomeBracket
df["IncomeBracket"] = pd.cut(
    df["Salary"],
    bins=[-np.inf, 5_000_000, 10_000_000, 15_000_000, 25_000_000, np.inf],
    labels=["<5m", "5-10m", "10-15m", "15-25m", ">=25m"]
)

df[['Salary', 'IncomeBracket']]
```

```python
#Trang thái khoản vay
import numpy as np

def classify_loan(row):
    if row["Trạng thái"] == "Nợ Xấu":
        return "Nợ xấu"

    elif row["Trạng thái"] == "Đang Vay" and row["LongestOverdue"] >= 1:
        return "Muộn"

    elif row["Trạng thái"] == "Kết thúc":
        return "Hoàn thành"

    else:
        return "Đang vay"

df["LoanStatus"] = df.apply(classify_loan, axis=1)
df[['Trạng thái','LongestOverdue', 'LoanStatus']]
```
