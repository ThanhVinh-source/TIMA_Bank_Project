```python
import pandas as pd
df = pd.read_csv('../data/tima_cleaned_data_v1.csv')
df
```

```python
#Create Dim_Customer
import pandas as pd

cust_cols = [
    "CardNumber","FullName","Gender","Birthday","Age","AgeGroup",
    "JobName","Salary","Hình thức cư trú", "TS_CREDIT_SCORE_V2","LowCreditScore", "CreditScoreGroup",
    "IncomeBracket","HasBadDebt"
]

dim_customer = df[cust_cols].drop_duplicates(subset=["CardNumber"]).reset_index(drop=True)
dim_customer.to_csv("../data/Dim_Fact Table/Dim_Customer.csv", index=False, encoding="utf-8-sig")
dim_customer.shape
dim_customer
```

```python
#Create Dim_Product
prod_cols = ["ProductCreditName","InterestPaymentType"]
dim_product = df[prod_cols].drop_duplicates(subset=["ProductCreditName"]).reset_index(drop=True)
dim_product.to_csv("../data/Dim_Fact Table/Dim_Product.csv", index=False, encoding="utf-8-sig")
dim_product.shape
dim_product
```

```python
#Create Dim_Date from application_date
# Do not touch df; only temporarily change the format to create the date range.
dates = pd.to_datetime(df["application_date"], errors="coerce")
dim_date = pd.DataFrame({"Date": pd.date_range(dates.min(), dates.max(), freq="D")})
dim_date["Year"]      = dim_date["Date"].dt.year
dim_date["Quarter"]   = "Q" + dim_date["Date"].dt.quarter.astype(str)
dim_date["MonthNum"]  = dim_date["Date"].dt.month
dim_date["MonthName"] = dim_date["Date"].dt.strftime("%b")
dim_date["MonthYear"] = dim_date["Date"].dt.strftime("%b %Y")
dim_date["Day"]       = dim_date["Date"].dt.day
dim_date["DayOfWeek"] = dim_date["Date"].dt.weekday + 1  # 1=Mon..7=Sun

dim_date.to_csv("../data/Dim_Fact Table/Dim_Date.csv", index=False, encoding="utf-8-sig")
dim_date.shape
dim_date
```

```python
#Create Dim_Geography
geo_cols = ["CityName","DistrictName"]
if geo_cols:
    dim_geography = df[geo_cols].drop_duplicates().reset_index(drop=True)
    dim_geography.to_csv("../data/Dim_Fact Table/Dim_Geography.csv", index=False, encoding="utf-8-sig")
    dim_geography.shape
dim_geography
```

```python
#Create Dim_Geography2
geo_cols = ["CityName"]
if geo_cols:
    dim_geography = df[geo_cols].drop_duplicates().reset_index(drop=True)
    dim_geography.to_csv("../data/Dim_Fact Table/Dim_Geography2.csv", index=False, encoding="utf-8-sig")
    dim_geography.shape
dim_geography
```

```python
#Create Fact_Loan table (containing information about loans)
fact_cols = [
    "LoanID","CardNumber","ProductCreditName","application_date","CityName", "Số tiền đăng ký vay ban đầu",
    "Tiền giải ngân","Cần Giải Ngân", "Tiền gốc còn lại", "Đã trả/khoản vay đã giải ngân","LongestOverdue","LoanTermMonths",
    "LoanToIncomeRatio","HasLatePayment","HasBadDebt","LoanStatus"
]

fact_loans = df[fact_cols].copy()

fact_loans.to_csv("../data/Dim_Fact Table/Fact_Loans.csv", index=False, encoding="utf-8-sig")
fact_loans.shape
fact_loans
```
