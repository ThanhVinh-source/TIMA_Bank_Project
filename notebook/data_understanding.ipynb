**Comments:**

- The data has 2383 rows and 49 columns.

- Some column data types need correction:

+, application_date: object --> date

+, Remaining amount: int64 --> float64

+, From Date: object --> datetime64

+, To Date: object --> datetime64

+, Gender: object --> Binary(0 - Female/1 - Male)

+, Birthday: object --> datetime64

+, Salary: object --> float64

+, CheckTime: object --> datetime64

+, HasBadDebt: int64 --> Binary (0 - No/1 - Yes)

+, HasLatePayment: int64 --> Binary (0 - No/1 - Yes)

```python
#Reformat data
#Synchronize monetary data (convert to the same currency, same float)
import pandas as pd
import numpy as np
money_cols = [
    "SoTienDKVayBanDau", "TienGiaiNgan", "SoTienConLai", "Salary", "Số tiền đăng ký vay ban đầu", "Tiền giải ngân", "Tiền gốc còn lại"
]
money_cols = [c for c in money_cols if c in df.columns]
#Convert USD to VND, keep VND unchanged
def convert_to_vnd(x, usd_to_vnd=26000): # USD -> VND exchange rate
    if pd.isna(x):
        return np.nan
    if isinstance(x, (int, float)):
        return float(x)
    s = str(x).lower().strip()
    if "$" in s or "usd" in s:
        # Take the numerical value and multiply it by the USD exchange rate.
        s = s.replace("$", "").replace("usd", "").strip()
        try:
            return float(s) * usd_to_vnd
        except:
            return np.nan
    # If there is no $, consider it VND
    try:
        return float(s)
    except:
        return np.nan

for col in money_cols:
    df[col] = df[col].apply(convert_to_vnd).astype(float)

df[money_cols]
```

```python
#Change the data format of date columns to datetime
def parse_ymd_date(series):
    return pd.to_datetime(
        series.astype("string").str.slice(0, 10),
        errors="coerce",
        format="%Y-%m-%d"
    )

date_cols = ["application_date", "FromDate", "ToDate", "Birthday"]
for col in date_cols:
    df[col] = parse_ymd_date(df[col])

df["CheckTime"] = pd.to_datetime(
    df["CheckTime"],
    errors="coerce",
    format="%Y-%m-%d %H:%M:%S"
)

df[date_cols + ["CheckTime"]]
```

```python
df[["FullName", "Gender"]]
```

```python
#Reformat some columns to Binary
import pandas as pd
import numpy as np

#Gender column (Gender: Male (Nam) --> 0, Female (Nữ) --> 1)
if "Gender" in df.columns:
    df["Gender"] = (
        df["Gender"]
        .astype(str)
        .str.strip()
        .str.lower()
        .replace({
            "nam": 0, "male": 0, "0":0,
            "nữ": 1,"female": 1, "1": 1
        })
    )
    df["Gender"] = pd.to_numeric(df["Gender"], errors="coerce").astype("Int64")

#HasBadDebt (1 = Yes, 0 = No)
if "HasBadDebt" in df.columns:
    df["HasBadDebt"] = (
        df["HasBadDebt"]
        .replace({"Có": 1, "Yes": 1, "yes": 1, True: 1, "1":1,
                  "Không": 0, "không": 0, "No": 0, "no": 0, False: 0, "0": 0})
    )
    df["HasBadDebt"] = pd.to_numeric(df["HasBadDebt"], errors="coerce").fillna(0).astype(int)

#HasLatePayment (1 = Yes, 0 = No)
if "HasLatePayment" in df.columns:
    df["HasLatePayment"] = (
        df["HasLatePayment"]
        .replace({"Có": 1, "co": 1, "có": 1, "Yes": 1, "yes": 1, True: 1,
                  "Không": 0, "không": 0, "No": 0, "no": 0, False: 0})
    )
    df["HasLatePayment"] = pd.to_numeric(df["HasLatePayment"], errors="coerce").fillna(0).astype(int)

df[["Gender", "HasBadDebt", "HasLatePayment"]]
```

```python
#In the NumberOfLoans column, if the value is less than 0, change it to 0
if "NumberOfLoans" in df.columns:
    df.loc[df["NumberOfLoans"] < 0, "NumberOfLoans"] = 0
```

**How ​​to handle missing data (missing values):**

- Remove columns with missing values ​​> 50%, including: ReceiveYourIncomeSalary (81.54%), WardName (72.01%)

- Drop the columns "SoTienDKVayBanDau", "TienGiaiNgan", "SoTienConLai" and use the "SoTienDKVayBanDau", "Tiền giải ngân", "SoTienConLai" columns to recalculate the value of the "SoTienConLai" column.

```python
#Drop 2 two columns with high missing value rates
drop_cols = [c for c in ["WardName", "ReceiveYourIncomeSalary"] if c in df.columns]
df = df.drop(columns=drop_cols)

print("Row :", df.shape[0])
print("Column :", df.shape[1])
```

```python
#Drop cột SoTienDKVayBanDau, TienGiaiNgan, SoTienConLai
drop_cols = [c for c in ["SoTienDKVayBanDau", "TienGiaiNgan", "SoTienConLai"] if c in df.columns]
df = df.drop(columns=drop_cols)

print("Row :", df.shape[0])
print("Column :", df.shape[1])
```

```python
#Fill Salary = Average Salary by Card Number
df["Salary"] = df["Salary"].fillna(
    df.groupby("CardNumber")["Salary"].transform("mean")
)

df['Salary']
```

```python
#If there are candidates with the same salary and from the same company, fill in the JobName for the remaining blank JobNames.
import pandas as pd
import numpy as np

if "NameCompany" in df.columns:
    df["NameCompany"] = (df["NameCompany"].astype(str)
                                         .str.strip()
                                         .replace({"": np.nan, "None": np.nan, "nan": np.nan}))

# Only fill in if JobName is missing and NameCompany & Salary are complete.
mask = df["JobName"].isna() & df["NameCompany"].notna() & df["Salary"].notna()

def safe_mode(s: pd.Series):
    m = s.mode(dropna=True)
    return m.iloc[0] if len(m) else np.nan

# Calculate mode of JobName for (NameCompany, Salary)
group_mode = (
    df.groupby(["NameCompany", "Salary"])["JobName"]
      .transform(safe_mode)
)

before_na = df["JobName"].isna().sum()
df.loc[mask, "JobName"] = df.loc[mask, "JobName"].fillna(group_mode[mask])
after_na = df["JobName"].isna().sum()
```

```python
# Fill Salary by mode of JobName
mode_by_job = df.groupby("JobName")["Salary"].agg(safe_mode)

df["Salary"] = df.apply(
        lambda row: mode_by_job[row["JobName"]] if pd.isna(row["Salary"]) and row["JobName"] in mode_by_job else row["Salary"],
        axis=1
    )
```

```python
for c in df.select_dtypes(include=["object", "category"]).columns:
    df[c] = df[c].fillna("Unknown")
```

```python
df[['JobName', 'Salary']]
```

```python
#For any remaining salary blanks or zeros, fill in the median.
df['Salary'] = df['Salary'].replace(0, np.nan).fillna(df['Salary'].median())
```

```python
#Change JobName 'Unknown' to 'Other'
df['JobName'] = df['JobName'].replace('Unknown', 'Khác')
```

```python
#Change Residence Type 'Unknown' to 'Other'
df['Hình thức cư trú'] = df['Hình thức cư trú'].replace('Unknown', 'Khác')
```
